Final function: concatenated log file should use all available rows in the original and restored, so that each replicate has the same number of columns in each log file, compared to other replicates, but some replicates might have more/less rows in each log file compared to other replicates. Here's the convo I had with Tiago about this reasoning:

**Me**:
I've had to restore some mcmc runs that didn't reach 100 million iterations, and in the process the final concatenated files end up having 100+ million iterations (varies, some are 150, some are 110, etc.). I can definitely cut the final concatenated logs to 100 million iterations across the board, but I also wanted to use all the data we have. So:
1. Do we want every run on every model w/ every replicate to have the same number of MCMC iterations/samples, in order to maintain consistency across the board as we compare results?
2. OR do we want to preserve the maximum number of MCMC iterations/samples available for each run, which means various models would have more samples than other models?


**Tiago**: 
The answer to your question is option number two. It doesn't really matter for all analyses to have the exact same number of generations. The goal is for us to have the posterior being sampled from a stationarity phase (some models will always take longer to get there relative to others, even if they have the same total number of generations). So in a sense, the total number of generations doesn't really matter... What matters is where your sampling is coming from. Also, all your parameter estimates will come from the parts of the posterior actually being sampled (e.g., if the sample at every 1000 generations, that means you will have 10,000 samples when running for 100 million generations). But you can always filter down your posterior to have exactly 10,000 samples across all model iterations no matter the total number of generations. In a sense, your total sample size matters much more than the total number of generations used.

**Note**: Note that the unnamed/empty header column ("") in the original marginal_rates.log is default renamed by Python to "Unnamed: 898". When I -combLogRJ (combine replicates' logs) with concatentaed marginal_rates.log files (have the "Unnamed: 898" column) and original marginal_rates.log files (have the unnamed/empty header column ""), this does not pose an issue. All that matters is that the files have the same # of columns


```python
# New function, response to the email above^

import pandas as pd
import numpy as np
import os

def concatenate_logs(original_log_prefix, restored_log_prefix, output_log_prefix):
    suffixes = ['_mcmc.log', '_ex_rates.log', '_sp_rates.log', '_per_species_rates.log', '_marginal_rates.log']
    
    def process_log_files(original_log_prefix, restored_log_prefix, suffix):
        original_log = original_log_prefix + suffix
        restored_log = restored_log_prefix + suffix
        output_log = output_log_prefix + '_CONCAT' + suffix

        # Check if the files exist
        if not os.path.exists(original_log) or not os.path.exists(restored_log):
            print(f"Skipping {suffix} because the file does not exist in the specified directory. Check path.")
            return None
        
        # Try reading the log files, handling parsing errors
        try:
            original = pd.read_csv(original_log, sep="\t", header=0)
        except pd.errors.ParserError as e:
            print(f"Skipping {suffix} because of a parsing error in the original log file: {e}. The log file may have rows with differing #'s of columns")
            return None
        
        try:
            restored = pd.read_csv(restored_log, sep="\t", header=0)
        except pd.errors.ParserError as e:
            print(f"Skipping {suffix} because of a parsing error in the restored log file: {e}. The log file may have rows with differing #'s of columns")
            return None
        
        # Strip any trailing or leading whitespace from the col names, so that no duplicate cols are created mistakenly 
        # If restored log has " 0.72" as a header, but original has "0.72", then a duplicate column will be created
        # This has only proven itself an issue with the _per_species_rates.log file, but it is good practice to do this for all log files
        original.columns = original.columns.str.strip()
        restored.columns = restored.columns.str.strip()
      
        # Drop the very first row in the restored log
        restored = restored.iloc[1:, :]
        concatenated = pd.concat([original, restored], axis=0)
        
        if concatenated.shape[0] == original.shape[0] + restored.shape[0]:
            # Keep all rows from the concatenated dataframe
            concatenated_final = concatenated
            
            # Update iteration columns based on the log type
            if suffix == '_mcmc.log':
                concatenated_final['it'] = np.arange(0, concatenated_final.shape[0] * 10000, 10000)
                if concatenated_final['it'].iloc[-1] != (concatenated_final.shape[0] - 1) * 10000:
                    raise ValueError(f"The last value in the 'it' column is not {(concatenated_final.shape[0] - 1) * 10000} for {suffix} logs.")
            elif suffix in ['_marginal_rates.log']:
                concatenated_final['it'] = np.arange(0, concatenated_final.shape[0] * 10000, 10000)
                if concatenated_final['it'].iloc[-1] != (concatenated_final.shape[0] - 1) * 10000:
                    raise ValueError(f"The last value in the 'it' column is not {(concatenated_final.shape[0] - 1) * 10000} for {suffix} logs.")
            
            concatenated_final.to_csv(output_log, sep="\t", index=False)
            print(f"\n Original ({suffix}): {original.shape[0]} rows, {original.shape[1]} columns \n",
                  f"Restored ({suffix}): {restored.shape[0]} rows, {restored.shape[1]} columns \n", 
                  f"Concatenated ({suffix}): {concatenated_final.shape[0]} rows, {concatenated_final.shape[1]} columns \n",)
            return concatenated_final
        else:
            raise ValueError(f"The concatenated log file is the wrong length for {suffix} logs.")
    
    results = {}
    for suffix in suffixes:
        result = process_log_files(original_log_prefix, restored_log_prefix, suffix)
        if result is not None: # Results would be none if the file didn't exist (if non-BDNN, with no _per_species_rates.log)
            results[suffix] = result
    return results


# Example Usage:
# Suppose you have the following log files:

# Original log files:
# ./reptilia/mcmc_no_predictors/A_bdnn/reptilia_pyrate_5_G_BDS_BDNN_4_2Tc_mcmc.log
# ./reptilia/mcmc_no_predictors/A_bdnn/reptilia_pyrate_5_G_BDS_BDNN_4_2Tc_ex_rates.log
# ./reptilia/mcmc_no_predictors/A_bdnn/reptilia_pyrate_5_G_BDS_BDNN_4_2Tc_sp_rates.log
# ./reptilia/mcmc_no_predictors/A_bdnn/reptilia_pyrate_5_G_BDS_BDNN_4_2Tc_per_species_rates.log
# Restored log files with the same names but in a different directory.
# You would call the function as follows:

# original_log_prefix = './reptilia/mcmc_no_predictors/A_bdnn/reptilia_pyrate_5_G_BDS_BDNN_4_2Tc'
# restored_log_prefix = './restored_logs/reptilia_pyrate_5_G_BDS_BDNN_4_2Tc'
# output_log_prefix = './output_logs/reptilia_pyrate_5_G_BDS_BDNN_4_2Tc'

# results = concatenate_logs(original_log_prefix, restored_log_prefix, output_log_prefix)
# This function will concatenate and process all the specified log files, save the resulting dataframes with the suffix _CONCAT.log, 
# and print out the relevant information for each log file. If any check fails, it will raise an error specifying which check failed.
```
